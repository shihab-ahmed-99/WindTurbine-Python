```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
import scipy.special as sc
```

```python
WS = pd.read_csv('windspeed.csv')
WS1 = WS.iloc[:,4:5]
WS1
```

```python
def weib(x,c,k):
  return (k/c)*(x/c)**(k-1)*np.exp(-(x/c)**k)

k = (math.sqrt(np.mean(abs(WS1-np.mean(WS1))))/np.mean(WS1))**-1.089

Gamma_F = math.exp(sc.gammaln(1+(1/k)))

c = np.mean(WS1)/Gamma_F

WiebDis = weib(WS1,c,k)
WiebDis.rename(columns = {'WS50M':'WeibDis'})
```

```python
plt.scatter(WS1,WiebDis, c = 'r',s = 1)
plt.xlabel('Wind Speed')
plt.ylabel('Probability Density')
plt.title('Weibull Distribution')
plt.show()
```
